```python
print("=== إحصائيات المبيعات ===")
sales_df[['SubTotal', 'TaxAmt', 'Freight', 'TotalDue']].describe()

print("\n=== إجمالي المبيعات الكلي ===")
total = sales_df['TotalDue'].sum()
print(f"${total:,.2f}")

print("\n=== أكبر 5 أوردرات ===")
sales_df.nlargest(5, 'TotalDue')[['SalesOrderID', 'OrderDate', 'TotalDue']]
 
```

```python
import matplotlib.pyplot as plt

# مبيعات كل سنة
sales_df['OrderYear'] = sales_df['OrderDate'].dt.year
yearly_sales = sales_df.groupby('OrderYear')['TotalDue'].sum()

yearly_sales.plot(kind='bar', figsize=(10,5), color='purple')
plt.title('إجمالي المبيعات السنوية - AdventureWorks')
plt.ylabel('المبيعات بالدولار')
plt.xlabel('السنة')
plt.show()

print("المبيعات بالسنة:")
yearly_sales
```

```python
query = """
SELECT TOP 5 p.FirstName, p.LastName, SUM(s.TotalDue) as TotalSpent
FROM Sales.SalesOrderHeader s
JOIN Sales.Customer c ON s.CustomerID = c.CustomerID
JOIN Person.Person p ON c.PersonID = p.BusinessEntityID
GROUP BY p.FirstName, p.LastName
ORDER BY TotalSpent DESC
"""

top_customers_sql = pd.read_sql(query, conn)
print("=== أكبر 5 عملاء - بـ SQL مباشر ===")
top_customers_sql
```

```python

df_top_products = pd.read_sql("""
    SELECT TOP 10
        p.Name AS ProductName,
        SUM(sod.LineTotal) AS TotalRevenue,
        SUM(sod.OrderQty) AS UnitsSold,
        AVG(sod.UnitPrice) AS AvgPrice
    FROM Sales.SalesOrderDetail sod
    JOIN Production.Product p ON sod.ProductID = p.ProductID
    GROUP BY p.Name
    ORDER BY TotalRevenue DESC
""", engine)

print("Top 10 Products by Revenue:")
print(df_top_products)

df_top_products.plot(kind='barh', x='ProductName', y='TotalRevenue', 
                     figsize=(12,6), color='#2E86AB', legend=False)
plt.title('Top 10 Products by Revenue - $123M AdventureWorks Dataset', fontsize=14, fontweight='bold')
plt.xlabel('Total Revenue ($)')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python

df_top_products = pd.read_sql("""
    SELECT TOP 10
        p.Name AS ProductName,
        SUM(sod.LineTotal) AS TotalRevenue,
        SUM(sod.OrderQty) AS UnitsSold
    FROM Sales.SalesOrderDetail sod
    JOIN Production.Product p ON sod.ProductID = p.ProductID
    GROUP BY p.Name
    ORDER BY TotalRevenue DESC
""", engine)

print("🏆 Top 10 Products by Revenue:")
print(df_top_products)


import matplotlib.pyplot as plt

df_top_products.plot(kind='barh', x='ProductName', y='TotalRevenue',
                     figsize=(12,7), color='#2E86AB', legend=False)
plt.title('Top 10 Products by Revenue - $123M Dataset', fontsize=14, fontweight='bold')
plt.xlabel('Total Revenue ($)')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

print(f"\n💰 أعلى منتج: {df_top_products['ProductName'].iloc[0]}")
print(f"💵 جاب كام: ${df_top_products['TotalRevenue'].iloc[0]:,.0f}")
```
